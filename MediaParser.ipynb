```python
import os
import subprocess
import json

comp_dir = "/content/drive/MyDrive/Colab_Transcriber/compressed"

print("🔍 Checking Compressed Video Durations:\n" + "="*60)
for root, dirs, files in os.walk(comp_dir):
    for f in sorted(files):
        if f.endswith(('.mp4', '.mkv', '.webm', '.mp3')):
            path = os.path.join(root, f)
            cmd = ['ffprobe', '-v', 'quiet', '-print_format', 'json', '-show_format', path]
            try:
                res = subprocess.run(cmd, capture_output=True, text=True, check=True)
                data = json.loads(res.stdout)
                dur = float(data['format']['duration'])
                size_mb = os.path.getsize(path) / (1024 * 1024)
                print(f"✅ {f}\n   ⏱️ Duration: {dur/60:.1f} min ({dur:.0f}s) | 💾 Size: {size_mb:.1f} MB\n")
            except Exception as e:
                print(f"❌ {f} - ERROR reading file: {e}\n")
```
